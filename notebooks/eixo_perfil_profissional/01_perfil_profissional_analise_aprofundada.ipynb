# Eixo 1 — Perfil profissional

## Pergunta de investigação

**Quem são os profissionais que participaram da pesquisa e quais características definem o perfil dessa amostra?**

### Objetivo

Caracterizar os participantes da pesquisa *Um Olhar de Quem Está no Campo*, considerando aspectos territoriais, demográficos, educacionais e de inserção profissional no setor de tecnologia.

### Variáveis analisadas

- Estado;
- Gênero;
- Faixa etária;
- Escolaridade;
- Área de atuação;
- Tempo de atuação em tecnologia.

### Estratégia de análise

A investigação foi organizada em quatro dimensões:

1. **Concentração territorial:** distribuição dos participantes por estado.
2. **Caracterização demográfica:** gênero e faixa etária.
3. **Formação profissional:** escolaridade.
4. **Inserção no setor:** áreas de atuação e tempo de experiência.

Foram realizados três cruzamentos exploratórios:

- Faixa etária × Tempo de atuação;
- Escolaridade × Área de atuação;
- Gênero × Área de atuação.

### Base e limites

A análise considera 54 respostas. A variável Área de atuação permite múltiplas seleções, totalizando 107 marcações em 15 categorias profissionais.

Os resultados caracterizam exclusivamente a amostra investigada, sem pretensão de representatividade estatística do conjunto dos profissionais de tecnologia.

# 03 — Investigação de cruzamentos pertinentes

*Análise conjunta das variáveis para identificar relações, padrões e evidências relevantes na composição do perfil profissional.*

```python
# Faixa etária × Tempo de atuação

cruzamento_idade_experiencia = pd.crosstab(
    df_perfil["Faixa Etária"], df_perfil[" Tempo de atuação na Área"]
)

display(cruzamento_idade_experiencia)
```

```python
# Distribuição percentual x faixa etária

cruzamento_idade_experiencia_pct = (
    pd.crosstab(
        df_perfil["Faixa Etária"],
        df_perfil[" Tempo de atuação na Área"],
        normalize="index",
    )
    * 100
).round(1)

display(cruzamento_idade_experiencia_pct)
```

```python
# Faixa etária × Tempo de atuação — Visualização

fig, ax = plt.subplots(figsize=(11, 6))

imagem = ax.imshow(
    cruzamento_idade_experiencia_pct, cmap="Blues", aspect="auto", vmin=0, vmax=100
)

# Rótulos dos eixos
ax.set_xticks(range(len(cruzamento_idade_experiencia_pct.columns)))
ax.set_xticklabels(cruzamento_idade_experiencia_pct.columns, rotation=35, ha="right")

# Total de participantes por faixa etária
total_por_idade = cruzamento_idade_experiencia.sum(axis=1)

# Identificação das faixas etárias com o respectivo n
rotulos_idade = [
    f"{idade} (n={total_por_idade.loc[idade]})"
    for idade in cruzamento_idade_experiencia.index
]

ax.set_yticks(range(len(rotulos_idade)))
ax.set_yticklabels(rotulos_idade)

# Percentual e quantidade de participantes em cada célula
for i in range(len(cruzamento_idade_experiencia_pct.index)):
    for j in range(len(cruzamento_idade_experiencia_pct.columns)):

        percentual = cruzamento_idade_experiencia_pct.iloc[i, j]
        quantidade = cruzamento_idade_experiencia.iloc[i, j]
        total = total_por_idade.iloc[i]

        ax.text(
            j,
            i,
            f"{percentual:.1f}%\n({quantidade}/{total})",
            ha="center",
            va="center",
            fontsize=8,
        )

ax.set_title("Tempo de atuação por faixa etária", fontsize=13, pad=15)
ax.set_xlabel("Tempo de atuação em tecnologia")
ax.set_ylabel("Faixa etária")

fig.colorbar(imagem, ax=ax, label="Participantes (%)")

plt.tight_layout()
plt.show()
```

### Leitura analítica

O cruzamento entre faixa etária e tempo de atuação evidencia diferentes configurações de experiência profissional entre os participantes da amostra.

Entre os participantes de **até 24 anos (n=9)**, observa-se uma concentração no intervalo de **1 a 3 anos de atuação**, que reúne 77,8% do grupo (7 participantes). Esse resultado indica uma presença predominante de profissionais em etapas iniciais de sua trajetória na área.

Na faixa de **45 a 54 anos (n=14)**, metade dos participantes (7) declarou mais de 15 anos de experiência. Entre aqueles com **55 anos ou mais (n=7)**, essa também é a categoria mais frequente, representando 42,9% do grupo (3 participantes), embora não corresponda à maioria.

Já entre os participantes de **35 a 44 anos (n=14)**, a distribuição é mais diversificada. A maior concentração está na faixa de 4 a 7 anos de atuação, com 35,7% (5 participantes), seguida por 8 a 15 anos, com 28,6% (4 participantes).

**Implicação metodológica:** os resultados permitem observar diferenças na distribuição do tempo de experiência entre as faixas etárias da amostra. Entretanto, é importante considerar os tamanhos dos grupos, especialmente nas categorias de 55 anos ou mais (n=7) e até 24 anos (n=9). Os percentuais devem ser interpretados em conjunto com as quantidades absolutas, sem extrapolar os resultados para o conjunto dos profissionais de tecnologia.

No mestrado, por exemplo, duas áreas concentram 50% das marcações cada, mas cada resultado corresponde a apenas uma marcação. No doutorado, as dez marcações estão distribuídas entre diferentes áreas.

**Implicação metodológica:** os percentuais representam a distribuição das marcações de áreas dentro de cada nível de escolaridade, e não a proporção de participantes de cada formação que atua em determinada área. Como a pergunta permitia múltiplas escolhas, uma pessoa pode contribuir com mais de uma marcação. Além disso, os tamanhos reduzidos de alguns grupos tornam os percentuais mais sensíveis a pequenas variações. Por isso, a interpretação deve considerar sempre os valores absolutos apresentados no gráfico e permanecer restrita à amostra investigada.

```python
# Gênero × Área de atuação

df_genero_area = (
    df_perfil[["Gênero", " Área de Atuação"]]
    .dropna(subset=[" Área de Atuação"])
    .assign(area=lambda df: df[" Área de Atuação"].str.split(","))
    .explode("area")
)

df_genero_area["area"] = df_genero_area["area"].str.strip()

cruzamento_genero_area = pd.crosstab(df_genero_area["Gênero"], df_genero_area["area"])

# Conferência
print(f"Total de marcações: {cruzamento_genero_area.values.sum()}")
print(f"Áreas identificadas: {len(cruzamento_genero_area.columns)}")

display(cruzamento_genero_area)
```

```python
# Distribuição percentual das áreas por gênero

cruzamento_genero_area_pct = (
    cruzamento_genero_area.div(cruzamento_genero_area.sum(axis=1), axis=0)
    .mul(100)
    .round(1)
)

display(cruzamento_genero_area_pct)
```

```python
# Gênero × Área de atuação — Visualização

# Excluir a categoria sem informação de gênero
dados_percentuais = cruzamento_genero_area_pct.drop(
    index="Prefiro não informar", errors="ignore"
)

dados_quantidade = cruzamento_genero_area.loc[dados_percentuais.index]

# Ordenar as áreas pelo total de marcações
ordem_areas = cruzamento_genero_area.sum(axis=0).sort_values(ascending=True).index

dados_percentuais = dados_percentuais[ordem_areas]
dados_quantidade = dados_quantidade[ordem_areas]

# Construção dos gráficos
fig, eixos = plt.subplots(nrows=2, ncols=1, figsize=(13, 12))

for ax, genero in zip(eixos, dados_percentuais.index):

    percentuais = dados_percentuais.loc[genero]
    quantidades = dados_quantidade.loc[genero]

    ax.barh(percentuais.index, percentuais.values)

    # Rótulos com percentual e quantidade de marcações
    for i, (percentual, quantidade) in enumerate(
        zip(percentuais.values, quantidades.values)
    ):
        ax.text(
            percentual + 0.4,
            i,
            f"{percentual:.1f}%  ({quantidade} marcações)",
            va="center",
            fontsize=8,
        )

    total_marcacoes = int(quantidades.sum())

    ax.set_title(f"{genero} — {total_marcacoes} marcações", loc="left", fontsize=11)

    ax.set_xlabel("Participação nas marcações (%)")
    ax.set_xlim(0, 45)
    ax.grid(axis="x", alpha=0.25)
    ax.set_axisbelow(True)

fig.suptitle("Distribuição das áreas de atuação por gênero", fontsize=14, y=1.01)

plt.tight_layout()
plt.show()
```

### Leitura analítica

O cruzamento evidencia semelhanças e diferenças na distribuição das áreas de atuação entre os grupos de gênero da amostra.

**Desenvolvimento de Software** concentra o maior número de marcações nos dois grupos: 21 entre os participantes do gênero masculino e 6 entre os participantes do gênero feminino. Considerando a distribuição percentual das marcações dentro de cada grupo, essa área representa 25,3% das marcações masculinas e 26,1% das femininas.

Entre as participantes do gênero feminino, além de Desenvolvimento de Software, destacam-se **Análise de Dados/Analytics (13,0%)**, **Gestão de Projetos em Tecnologia (8,7%)**, **Arquitetura de Soluções (8,7%)**, **UX/UI/Design de Produto (8,7%)** e **Engenharia de Dados (8,7%)**.

Entre os participantes do gênero masculino, aparecem com maior participação, depois de Desenvolvimento de Software, **Inteligência Artificial/Machine Learning (15,7%)**, **Análise de Dados/Analytics (13,3%)** e **Gestão de Projetos em Tecnologia (10,8%)**.

**Implicação metodológica:** os percentuais representam a distribuição das marcações de áreas dentro de cada grupo de gênero, e não a proporção de pessoas de cada gênero que atua em determinada área. Como a questão permitia múltiplas seleções, cada participante podia contribuir com mais de uma marcação. Além disso, os grupos apresentam tamanhos diferentes, tanto em número de participantes quanto em número de marcações. Portanto, os resultados permitem observar diferenças na composição das áreas assinaladas, mas não permitem concluir, isoladamente, que determinada área tenha maior presença de mulheres ou homens no mercado de tecnologia.

Nas faixas de 45 a 54 anos e de 55 anos ou mais, a categoria mais frequente é a de profissionais com mais de 15 anos de experiência, correspondendo a 50% e 42,9%, respectivamente. No grupo de 55 anos ou mais, essa proporção não representa a maioria dos participantes.

Na relação entre escolaridade e área de atuação, Desenvolvimento de Software concentra a maior participação entre as marcações de profissionais com graduação (35,6%). Entre as marcações de participantes com pós-graduação, destacam-se Inteligência Artificial/Machine Learning (17,1%) e Análise de Dados/Analytics (14,3%). Na formação técnica, Análise de Dados/Analytics e Desenvolvimento de Software representam, cada uma, 33,3% das marcações. Esses resultados devem ser observados com cautela, especialmente nos níveis de escolaridade com poucos participantes.

Já o cruzamento entre gênero e área de atuação mostra que Desenvolvimento de Software concentra o maior número de marcações nos dois grupos. Quando observada a distribuição percentual das marcações dentro de cada grupo, a participação é próxima: 25,3% entre homens e 26,1% entre mulheres. Outras diferenças aparecem na distribuição das áreas, mas os resultados não permitem concluir que determinados campos sejam predominantemente masculinos ou femininos.

### Síntese do perfil observado

Em conjunto, os resultados caracterizam uma amostra predominantemente catarinense e masculina, composta majoritariamente por profissionais com graduação ou pós-graduação. Ao mesmo tempo, reúne participantes de diferentes faixas etárias, níveis de experiência e campos de atuação.

Os cruzamentos acrescentam outras perspectivas à caracterização inicial, mostrando como as marcações se distribuem entre os grupos analisados. Essas informações ajudam a compreender a composição interna da amostra, sem estabelecer relações causais ou representar necessariamente a realidade de todo o mercado de tecnologia.

### Limitações e cuidados de interpretação

Os achados devem ser compreendidos dentro dos limites da amostra de 54 participantes, que não possui pretensão de representatividade estatística do conjunto dos profissionais de tecnologia no Brasil.

A concentração territorial em Santa Catarina restringe a abrangência geográfica das observações. Além disso, algumas categorias apresentam poucos participantes, o que exige cautela na interpretação dos cruzamentos, especialmente quando envolvem grupos reduzidos.

A variável Área de Atuação permitia múltiplas seleções. Por isso, suas frequências e percentuais representam marcações, e não necessariamente profissionais exclusivos em cada campo. Os cruzamentos envolvendo essa variável devem ser interpretados considerando essa característica.

Conclui-se que a análise aprofundada deste primeiro eixo estabelece uma caracterização dos profissionais que participaram da pesquisa e oferece uma base contextual para a interpretação dos resultados dos demais eixos: Jornada, Realidade e Olhando para frente.
